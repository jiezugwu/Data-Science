## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
A fixed rule works when one or two inputs decide the outcome and mean the same thing everywhere, but the data breaks both conditions. A signal's meaning depends on context: strong engagement at position 2 suggests holding or slipping, while the same engagement at position 15 suggests climbing, and a 20-day-old page behaves differently from a 400-day-old one at the same position. The right thresholds are unknown and likely differ by slice, GA4 metrics are missing for 26% of pages, 63% of pages have no clicks, and 54 clients behave differently, so a rule would need a branch for every combination. Movement also comes from many weak signals adding up, and noisy low-impression positions (like a page with 53 impressions) would make a rule react to randomness. ML can learn these interactions, thresholds and noise levels from the data. It still has to prove itself by beating simple baselines like "nothing changes" on held-out clients, and if it can't, the simpler rule is the better choice.
