## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*
The main limitation is that decline_recovery contains only pages that had already dropped 30% or more in impressions, so the population was selected on the outcome. Every row is a decliner, which means this slice can't show what separates pages that decline from pages that don't, only what is associated with steeper loss among pages that already fell. Any signal we find may also describe how pages got into the slice, not how severe their drop became. The findings therefore apply only to this population, and they can't be stated as general drivers of decline or as anything that predicts recovery. ai_opportunity has the mix of down, stable and up pages, but it shares no pages with the two tables we're using, so it can't fill the gap by joining.
